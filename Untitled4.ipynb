```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
from collections import Counter

# Get uploaded image name
image_path = list(uploaded.keys())[0]

# Load YOLO model
model = YOLO("yolov8n.pt")

# Read image
image = cv2.imread(image_path)

# Safety check
if image is None:
    print("Error: Image not loaded properly")
else:
    # Run detection
    results = model(image)

    object_counts = Counter()

    for result in results:
        boxes = result.boxes

        for box in boxes:
            cls = int(box.cls[0])
            label = model.names[cls]

            # Count objects
            object_counts[label] += 1

            # Draw bounding box
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            cv2.rectangle(image, (x1, y1), (x2, y2), (0,255,0), 2)
            cv2.putText(image, label, (x1, y1-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)

    # Convert BGR → RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # SHOW OUTPUT IMAGE ✅
    plt.imshow(image_rgb)
    plt.axis('off')
    plt.title("YOLO Object Detection Output")
    plt.show()

    # PRINT COUNTS (IMPORTANT PROOF)
    print("\nDetected Objects:")
    for obj, count in object_counts.items():
        print(f"{obj}: {count}")

    # PIE CHART ✅
    if len(object_counts) > 0:
        plt.figure()
        plt.pie(object_counts.values(), labels=object_counts.keys(), autopct='%1.1f%%')
        plt.title("Object Distribution")
        plt.show()
    else:
        print("No objects detected")
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# Load model
model = YOLO("yolov8n.pt")

# Get uploaded image
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Check image
if image is None:
    print("❌ Image not loaded")
else:
    print("✅ Image loaded")

# Run detection (NORMAL)
results = model.predict(source=image, conf=0.25)

# Draw boxes
detected = False

for r in results:
    if r.boxes is not None and len(r.boxes) > 0:
        detected = True
        for box in r.boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            cls = int(box.cls[0])
            label = model.names[cls]

            cv2.rectangle(image, (x1,y1),(x2,y2),(0,255,0),2)
            cv2.putText(image, label, (x1,y1-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0),2)

# Show image
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image_rgb)
plt.axis('off')

if detected:
    print("✅ Objects detected")
    plt.title("Objects Detected")
else:
    print("❌ No objects detected")
    plt.title("No Objects Detected")

plt.show()
```
